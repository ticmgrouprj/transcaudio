```python

# === Instalar bibliotecas necessárias ===
!pip install -q openai-whisper gdown pydub
!apt-get -q install ffmpeg

# === Importar bibliotecas ===
import os
import math
import gdown
import whisper
from pydub import AudioSegment

# === Função para formatar tempo no modelo [hh:mm:ss] ===
def format_ts(seconds):
    h, rem = divmod(int(seconds), 3600)
    m, s = divmod(rem, 60)
    return f"[{h:02}:{m:02}:{s:02}]"

# === Função principal de transcrição ===
def transcrever_audio_do_drive(link_drive, modelo="medium", duracao_min=10):
    print("🔗 Extraindo ID do link...")
    if "id=" in link_drive:
        file_id = link_drive.split("id=")[-1].split("&")[0]
    elif "/d/" in link_drive:
        file_id = link_drive.split("/d/")[-1].split("/")[0]
    else:
        raise ValueError("❌ Link inválido do Google Drive.")

    url_direct = f"https://drive.google.com/uc?id={file_id}"

    print("📥 Baixando arquivo de áudio...")
    local_audio = "audio.mp3"
    gdown.download(url_direct, local_audio, quiet=False)

    print("🔊 Carregando áudio...")
    audio = AudioSegment.from_file(local_audio)
    chunk_ms = duracao_min * 60 * 1000
    total_chunks = math.ceil(len(audio) / chunk_ms)

    print("🧠 Carregando modelo Whisper...")
    model = whisper.load_model(modelo)

    print("\n📝 Iniciando transcrição...\n")
    for i in range(total_chunks):
        ini_ms = i * chunk_ms
        fim_ms = min((i + 1) * chunk_ms, len(audio))
        chunk = audio[ini_ms:fim_ms]
        chunk.export("temp_chunk.mp3", format="mp3")

        print(f"\n🔹 Trecho {i+1}/{total_chunks} - Início: {format_ts(ini_ms // 1000)}")
        result = model.transcribe("temp_chunk.mp3", language="pt")

        for seg in result["segments"]:
            ini = ini_ms / 1000 + seg["start"]
            print(f"{format_ts(ini)} {seg['text'].strip()}")

    print("\n✅ Transcrição finalizada!")

# === Solicita o link ao usuário ===
link_usuario = input("📎 Cole o link do Google Drive com o áudio: ")
transcrever_audio_do_drive(link_usuario)


```
